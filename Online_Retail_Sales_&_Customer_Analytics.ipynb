# ***Online Retail Sales & Customer Analytics***

**Descripción ejecutiva del proyecto**

El proyecto analiza el comportamiento comercial de una empresa de retail online a partir de una base de 541.909 registros de transacciones, con información sobre pedidos, productos, cantidades, precios, clientes, fechas y países.

El objetivo fue transformar una base transaccional con problemas de calidad en un conjunto de datos confiable para analizar:

*   Evolución de las ventas.
*   Comportamiento de los clientes.
*   Productos con mayor demanda.
*   Productos con mayor generación de ingresos.
*   Distribución geográfica de las ventas.
*   Evolución mensual de los ingresos.
*   Calidad y consistencia de los datos.

1. Comenzamos con la base de datos de **541910 registros** en la hoja **Online Retail**

2. ***Limpieza y calidad de datos***

La base original tenía **541.909 registros**. Después del proceso de limpieza quedaron **539.388 registros**, es decir, se eliminaron **2.521 registros**. Pero no todos los registros con valores problemáticos fueron tratados de la misma manera.

Se eliminaron registros que presentaban problemas como:
* Cantidades iguales a cero.
* Cantidades negativas sin información suficiente para interpretarlas.
* Registros sin descripción de producto.

Se realizo el cambio respectivo a **9287 registros** que contaban con cantidades negativas que muestran un mal registro pero contaban con precio, descripcion y cliente esta cambio se dio con la columna calculada **cantidad corregida** que es: ABS(Quantity) valor absoluto de la cantidad.

Tambien se calculo una nueva columna **Precio total** que es:

Precio Total = Cantidad corregida × Precio Unitario

Con estos respectivos cambios en los datos, se dio inicio al analisis, los datos corregidos estan en la hoja **DATOS LIMPIOS** y la descripcion de los registros eliminados y corregidos estan en la hoja **README**

3. **Problema de negocio**

La empresa cuenta con una gran cantidad de información transaccional, pero necesita convertir esos datos en información confiable que permita entender el desempeño comercial, identificar clientes y productos relevantes y detectar oportunidades de mejora en la calidad de los datos.

**Preguntas del negocio:**

**Ventas**

* ¿Cuánto vende la empresa?
* ¿Cómo evolucionan las ventas a lo largo del tiempo?
* ¿Qué meses generan mayores ingresos?
* ¿Qué países generan mayor facturación?

**Productos**

* ¿Qué productos tienen mayor volumen de unidades vendidas?
* ¿Qué productos generan mayores ingresos?
* ¿Qué productos concentran la mayor cantidad de transacciones?

**Clientes**

* ¿Qué clientes tienen mayor actividad?
* ¿Qué clientes generan mayores ingresos?
* ¿Cuántas transacciones no pueden asociarse a un cliente identificado?
